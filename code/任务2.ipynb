任务 2.1 统计不同登记年份不同产品来源的特医食品获批量，绘制双折线图，并在报 告中对结果进行必要分析。

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['font.sans-serif']=['SimHei'] # 中文支持
plt.rcParams['axes.unicode_minus']=False
```

```python
import pandas as pd


file_path = 'result2.xlsx'
df = pd.read_excel(file_path)

df.head()
```

```python
import matplotlib.pyplot as plt

# 统计不同登记年份不同产品来源的特医食品获批量
approval_counts = df.groupby(['登记年份', '产品来源']).size().unstack(fill_value=0)

# 绘制双折线图
plt.figure(figsize=(10, 6))
plt.plot(approval_counts.index, approval_counts['国产产品'], marker='o', label='国产产品')
plt.plot(approval_counts.index, approval_counts['进口产品'], marker='o', label='进口产品')
plt.title('不同登记年份不同产品来源的特医食品获批量')
plt.xlabel('登记年份')
plt.ylabel('获批量')
plt.legend()
plt.grid(True)
plt.xticks(approval_counts.index, rotation=45)
plt.tight_layout()
plt.show()
```


任务 2.2 根据特医食品产品来源与适用人群类别绘制内层为饼图的旭日图，其中内层 表示适用人群类别，外层表示不同适用人群类别的产品来源分布，并在报告中对结果进行 必要分析。

```python
import matplotlib.pyplot as plt

# 统计特医食品产品来源与适用人群类别的分布情况
sunburst_data = df.groupby(['适用人群类别', '产品来源']).size().unstack(fill_value=0)

# 绘制内层为饼图的旭日图
plt.figure(figsize=(10, 8))
inner_labels = sunburst_data.index
outer_labels = ['{} - {}'.format(inner, outer) for inner in inner_labels for outer in sunburst_data.columns]

# 内层数据（适用人群类别的总和）
inner_sizes = sunburst_data.sum(axis=1)

# 外层数据（各类别下不同产品来源的数量）
outer_sizes = sunburst_data.values.flatten()

# 颜色设置
colors = plt.cm.Paired(range(len(inner_sizes) + len(outer_sizes)))

# 绘制饼图
plt.pie(inner_sizes, labels=inner_labels, colors=colors[:len(inner_sizes)], radius=1, wedgeprops=dict(width=0.3))
plt.pie(outer_sizes, labels=outer_labels, colors=colors[len(inner_sizes):], radius=1.3, wedgeprops=dict(width=0.3))

plt.title('特医食品产品来源与适用人群类别的旭日图')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# 绘制内层为饼图的旭日图，并将文字放入图内
plt.figure(figsize=(10, 8))
inner_labels = sunburst_data.index
outer_labels = ['{} - {}'.format(inner, outer) for inner in inner_labels for outer in sunburst_data.columns]

# 内层数据（适用人群类别的总和）
inner_sizes = sunburst_data.sum(axis=1)
inner_explode = [0.05] * len(inner_sizes)  # 适当调整内层饼图的分隔以更清晰显示

# 外层数据（各类别下不同产品来源的数量）
outer_sizes = sunburst_data.values.flatten()
outer_explode = [0.05] * len(outer_sizes)  # 适当调整外层饼图的分隔以更清晰显示

# 颜色设置
colors = plt.cm.Paired(range(len(inner_sizes) + len(outer_sizes)))

# 绘制内层饼图
wedges1, texts1, autotexts1 = plt.pie(
    inner_sizes, labels=inner_labels, colors=colors[:len(inner_sizes)],
    radius=1, wedgeprops=dict(width=0.3, edgecolor='w'), explode=inner_explode,
    autopct='%.1f%%', pctdistance=0.75
)

# 绘制外层饼图
wedges2, texts2, autotexts2 = plt.pie(
    outer_sizes, labels=outer_labels, colors=colors[len(inner_sizes):],
    radius=1.3, wedgeprops=dict(width=0.3, edgecolor='w'), explode=outer_explode,
    autopct='%.1f%%', pctdistance=0.85
)

# 优化文字显示位置
for text in autotexts1 + autotexts2:
    text.set_color('black')
    text.set_fontsize(10)

plt.title('特医食品产品来源与适用人群类别的旭日图')
plt.tight_layout()
plt.show()
```

任务 2.3 统计不同产品类别的特医食品获批量，按获批量进行降序排列，绘制柱状 图，x 轴为产品类别，y 轴为获批量，并在报告中对结果进行必要分析。

```python
# 统计不同产品类别的特医食品获批量，并按降序排列
product_approval_counts = df['产品类别'].value_counts().sort_values(ascending=False)

# 绘制柱状图
plt.figure(figsize=(12, 6))
plt.bar(product_approval_counts.index, product_approval_counts.values)
plt.title('不同产品类别的特医食品获批量')
plt.xlabel('产品类别')
plt.ylabel('获批量')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

任务 2.4 在同一坐标系中，分别用不同颜色绘制 182 款特医食品脂肪和蛋白质含量的 频数分布直方图，并在报告中对结果进行必要分析。

```python
file_path = 'result1.xlsx'
df2 = pd.read_excel(file_path)

df2.head()


# 提取脂肪和蛋白质含量的列，并移除缺失值
fat_protein_data = df2[['脂肪(g)', '蛋白质(g)']].dropna()

# 绘制脂肪和蛋白质含量的频数分布直方图
plt.figure(figsize=(12, 6))

plt.hist(fat_protein_data['蛋白质(g)'], alpha=0.7, label='蛋白质(g)', color='orange')
plt.hist(fat_protein_data['脂肪(g)'], alpha=0.7, label='脂肪(g)', color='blue')
plt.title('182款特医食品脂肪和蛋白质含量的频数分布')
plt.xlabel('含量（g/100kj）')
plt.ylabel('频数')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```


任务 2.5 根据 182 款特医食品的“适用人群”绘制词云图，并在报告中分析特医食品 适用人群特征。

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# 使用 Windows 系统上的中文字体路径
font_path = 'C:\\Windows\\Fonts\\SimHei.ttf'

# 提取适用人群的文本数据
applicable_population = df['适用人群'].dropna().astype(str).tolist()
text = ' '.join(applicable_population)

# 生成词云图
wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white',
    font_path=font_path
).generate(text)

# 绘制词云图
plt.figure(figsize=(10, 6))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('182款特医食品的适用人群词云图')
plt.show()
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt
import jieba
import pandas as pd

# 中文显示支持
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 使用 Windows 系统上的中文字体路径
font_path = 'C:\\Windows\\Fonts\\SimHei.ttf'

# 提取适用人群的文本数据
applicable_population = df['适用人群'].dropna().astype(str).tolist()
text = ' '.join(applicable_population)

# 使用 jieba 分词
seg_list = jieba.cut(text, cut_all=False)
seg_text = ' '.join(seg_list)

# 手动设置停用词列表
stopwords = set(['的', '和', '是', '在', '了', '有', '为', '与', '以及', '可以', '等', '其', '对', '中', '于', '不', '使用'
                '需要','岁','以上','人群',' 需要',])

# 生成词云图
wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white',
    font_path=font_path,
    stopwords=stopwords
).generate(seg_text)

# 绘制词云图
plt.figure(figsize=(10, 6))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('182款特医食品的适用人群词云图')
plt.show()
```
